````python
# Load env variables
from dotenv import load_dotenv

load_dotenv()
````

````python
# Create an API client
from anthropic import Anthropic
client = Anthropic()
model = "claude-sonnet-4-5"
````

````python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None, stop=None):
    params = {
        "model":model,
        "max_tokens":1000,
        "messages":messages,
    }

    if system:
        params["system"] = system

    if stop:
        params["stop_sequences"] = stop

    message = client.messages.create(**params)
    return message.content[0].text
````

````python
# Make a starting list of messages

messages = []

add_user_message(messages, "Generate a very short event bridge rule as json")

# Prefill assistant with opening markdown fence

add_assistant_message(messages, "```json")

# Get answer and stop before closing fence

answer = chat(messages, stop=["```"])

print(answer)
````

````python
messages = []

prompt = """
Generate three different sample AWS CLI commands. Each should be very short.
"""

add_user_message(messages, prompt)
add_assistant_message(messages, "Here are all three commands in a single block without any comments:\n```bash")

text = chat(messages, stop=["```"])
text.strip()
````
